```python
#1)simple route to greet the user
from flask import Flask
app= Flask(__name__)
@app.route('/')
def home():
    return "hello bobby darlingg!!"

if __name__ == '___main__':
    app.run(debug=False, use_reloader=False)
```

```python
#2)ROUTE WITH USER INPUT ANS COMPUTATION
from flask import Flask, request, render_template_string
app = Flask(__name__)

#captures string name fromt he url
@app.route('/greet/<name>')
def greet_user(name):
    return render_template_string('<h1>Hello, {(name)}!</h1>', name=name)

@app.route('/square/<int:number>')
def square(number):
    result = number ** 2
    return render_template_string('<h1>{{number}} squared is {{result}}</h1>', number=number, result=result)

if __name__ == "__main__":
    app.run(debug=False)
```

```python
#4)ROUTE WITH POST METHOD
from flask import Flask, request, render_template_string
app = Flask(__name__)

@app.route('/submit', methods=['GET', 'POST'])
def submit():
    if request.method=='POST':
        message = request.form.get('message', 'No message')
        return render_template_string("<h1>you submitted : {{message}}</h1>", message=message)
    else:
        return ''' 
            <form method="post" action="/submit">
                <input type="text" name="message" />
            </form>
  
        '''

if __name__ == "__main__":
    app.run(debug=False, use_reloader=False)
```

```python
from flask import Flask, request, jsonify
app= Flask(__name__)

@app.route('/submit', methods=['GET', 'POST'])
def submit():
    if request.method =='POST':
        data = request.get_json()
        message = data.get('message', 'No message')
        return jsonify({'response': f"you submitted: {message}"})
    else:
        return ''' 
            <form method="post" action="/submit">
                <input type="text" name="message" />
            </form>
        '''
    
if __name__ == "__main__":
    app.run(debug=False, use_reloader=False)
```
