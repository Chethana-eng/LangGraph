```python
import sqlite3
import pandas as pd
```

```python
connection = sqlite3.connect(':memory:')

cursor = connection.cursor()
```

```python
create_table_query = '''
create table student(
    id integer primary key autoincrement,
    name text,
    age integer,
    mlmarks integer,
    dlmarks integer
);
'''
cursor.execute(create_table_query)
print("table created")
```

```python
insert_data= '''
insert into student(name, age, mlmarks, dlmarks)
values
('bob', 20, 90,70),
('riya', 21, 50,100),
('mintu', 25, 90,40);
'''
cursor.execute(insert_data)
connection.commit()
print("data inserted")
```

```python
query = "select * from student;"
df = pd.read_sql(query, connection)
df
```

```python
update_query = "update student set mlmarks=100 where name='riya';"
cursor.execute(update_query)
connection.commit()
```
