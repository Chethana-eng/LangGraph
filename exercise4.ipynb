```python
from typing import TypedDict
from langgraph.graph import StateGraph, START, END
```

```python
class AgentState(TypedDict):
    number1: int
    operation1: str
    number2: int
    finalnum1: int
    number3: int
    operation2: str
    number4: int
    finalnum2: int
```

```python
def adder1(state: AgentState) ->AgentState:
    """adds 2 numbers"""

    state["finalnum1"] = state['number1'] + state['number2']

    return state

def sub1(state: AgentState) ->AgentState:
    """subtracts 2 numbers"""

    state["finalnum1"] = state['number1'] - state['number2']

    return state

def decide_node1(state: AgentState) ->AgentState:
    """this node will select the next node int the graph"""
    if state["operation1"]=="+":
        return "addition_operation1"
    if state["operation2"]=="-":
        return "subtraction_operation1"
    


def adder2(state: AgentState) ->AgentState:
    """adds 2 numbers"""

    state["finalnum2"] = state['number3'] + state['number4']

    return state

def sub2(state: AgentState) ->AgentState:
    """subtracts 2 numbers"""

    state["finalnum2"] = state['number3'] - state['number4']

    return state

def decide_node2(state: AgentState) ->AgentState:
    """this node will select the next node int the graph"""
    if state["operation2"]=="+":
        return "addition_operation2"
    if state["operation2"]=="-":
        return "subtraction_operation2"
```
